```python
pairs = []
for i in range(len(resume_strings)):
    job_idx = i // 10
    pairs.append({
        "job": job_strings[job_idx],
        "resume": resume_strings[i],
        "label": labels_mapping.get(resume_strings[i][:8], 0.5)
    })
```

```python
from transformers import BertTokenizer

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

train_inputs = []
train_labels = []

for pair in pairs:
    encoded = tokenizer(
        text=pair["job"],
        text_pair=pair["resume"],
        padding='max_length',
        truncation=True,
        max_length=512,
        return_tensors='tf'
    )
    train_inputs.append(encoded)
    train_labels.append(pair["label"])

print(f"Training samples: {len(train_inputs)}")

```

```python
print(type(train_inputs))

#normalizing train_labels

train_labels = np.array(train_labels)/100.0
train_labels
```

```python
for x in train_inputs:
  print(x["attention_mask"])
```

```python
# attention_mask = [1, 1, 0, 0, 0, ...]

# BERT formula:
# attention_score = softmax(QK^T + (mask × -10000))

# mask=0 → -10000 add → softmax makes it 0
# mask=1 → 0 add → normal calculation

# Suppose BERT output:
# bert_output = {
#     'last_hidden_state': [
        # Batch 0: 3 tokens, 4 features each
#         [
#             [0.1, 0.2, 0.3, 0.4],   # Token 0: [CLS] embedding
#             [0.5, 0.6, 0.7, 0.8],   # Token 1: First word
#             [0.9, 1.0, 1.1, 1.2]    # Token 2: Second word
#         ]
#     ]
# }
# Shape: (1 batch, 3 tokens, 4 features)

# [:, 0, :] =
# :   = All batches (batch 0, batch 1, ...)
# 0   = First token ([CLS] token)
# :   = All 768 features
#BERT Make CLS embed we just extract it(summary) then send to output layer to get answer

# tf.constant() = "Convert to TensorFlow array"
```

```python
import tensorflow as tf
import numpy as np
from transformers import TFBertModel

input_ids = tf.keras.layers.Input(shape=(512,), dtype=tf.int32, name='input_ids')
attention_mask = tf.keras.layers.Input(shape=(512,), dtype=tf.int32, name='attention_mask')
token_type_ids = tf.keras.layers.Input(shape=(512,), dtype=tf.int32, name='token_type_ids')

# BERT model
bert = TFBertModel.from_pretrained('bert-base-uncased', from_pt=True)
bert_output = bert({'input_ids': input_ids,
                    'attention_mask': attention_mask,
                    'token_type_ids': token_type_ids})

# Regression head
cls_embedding = bert_output.last_hidden_state[:, 0, :]
dense_layer = tf.keras.layers.Dense(1, activation='sigmoid')
outputs = dense_layer(cls_embedding)

model = tf.keras.Model(inputs=[input_ids, attention_mask, token_type_ids],
                       outputs=outputs)
```

```python
model.summary()
```

```python

X_train = {
    'input_ids': tf.concat([x['input_ids'] for x in train_inputs], axis=0),
    'attention_mask': tf.concat([x['attention_mask'] for x in train_inputs], axis=0),
    'token_type_ids': tf.concat([x['token_type_ids'] for x in train_inputs], axis=0)
    }

y_train = tf.constant(train_labels, dtype=tf.float32)


print(type(X_train["input_ids"]))
print(type(y_train))
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=3e-5),
    loss='mse',
    metrics=['mae']
)

print("Training starting...")
history = model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=1,
    verbose=1
)

print("Training complete!")
```

```python
loss, mae = model.evaluate(X_train, y_train, verbose=2)
print(f"Evaluation:")
print(f"Loss (MSE): {loss:.6f}")
print(f"MAE: {mae:.6f}")
```

```python
parsers_code = '''import os
import zipfile
import re
import PyPDF2

def parse_odt(file_path):
    try:
        from odf.opendocument import load
        import odf.text as odf_text
        import odf.teletype as teletype
        doc = load(file_path)
        all_text = []
        for paragraph in doc.getElementsByType(odf_text.P):
            all_text.append(teletype.extractText(paragraph))
        text = "\\n".join(all_text)
        return clean_text(text)
    except Exception as e:
        raise Exception(f"ODT error: {str(e)}")

def parse_pdf(file_path):
    text = ""
    try:
        with open(file_path, 'rb') as file:
            pdf_reader = PyPDF2.PdfReader(file)
            for page in pdf_reader.pages:
                page_text = page.extract_text()
                if page_text:
                    text += page_text + "\\n"
    except Exception as e:
        print(f"PDF error: {str(e)}")
    return clean_text(text)

def parse_docx(file_path):
    try:
        from docx import Document
        doc = Document(file_path)
        full_text = []
        for paragraph in doc.paragraphs:
            if paragraph.text.strip():
                full_text.append(paragraph.text)
        for table in doc.tables:
            for row in table.rows:
                row_text = []
                for cell in row.cells:
                    if cell.text.strip():
                        row_text.append(cell.text.strip())
                if row_text:
                    full_text.append(" | ".join(row_text))
        text = "\\n".join(full_text)
        return clean_text(text)
    except Exception as e:
        raise Exception(f"DOCX error: {str(e)}")

def parse_txt(file_path):
    try:
        with open(file_path, 'r', encoding='utf-8') as file:
            text = file.read()
    except UnicodeDecodeError:
        try:
            with open(file_path, 'r', encoding='latin-1') as file:
                text = file.read()
        except Exception as e:
            raise Exception(f"TXT error: {str(e)}")
    except Exception as e:
        raise Exception(f"TXT error: {str(e)}")
    return clean_text(text)

def parse_file(file_path):
    if not os.path.exists(file_path):
        raise FileNotFoundError(f"File not found: {file_path}")
    if os.path.isdir(file_path):
        raise ValueError(f"'{file_path}' is a directory, not a file!")
    file_ext = os.path.splitext(file_path)[1].lower()
    if file_ext == '.odt':
        return parse_odt(file_path)
    elif file_ext == '.pdf':
        return parse_pdf(file_path)
    elif file_ext == '.docx':
        return parse_docx(file_path)
    elif file_ext == '.txt':
        return parse_txt(file_path)
    else:
        raise ValueError(f"Unsupported format: {file_ext}")

def clean_text(text):
    if not text or not isinstance(text, str):
        return ""
    text = text.replace('\\xa0', ' ')
    text = text.replace('\\t', ' ')
    text = text.replace('\\r\\n', '\\n')
    text = text.replace('\\r', '\\n')
    lines = text.split('\\n')
    cleaned_lines = [line.strip() for line in lines if line.strip()]
    text = '\\n'.join(cleaned_lines)
    text = re.sub(r'\\s+', ' ', text)
    return text.strip()

def find_all_files(directory):
    all_files = []
    for root, dirs, files in os.walk(directory):
        for file in files:
            if not file.startswith('.') and file not in ['mimetype', 'settings.xml', 'styles.xml', 'meta.xml', 'content.xml', 'manifest.rdf', 'manifest.xml']:
                full_path = os.path.join(root, file)
                all_files.append(full_path)
    return all_files
'''

with open('parsers.py', 'w') as f:
    f.write(parsers_code)
print("parsers.py created")
```

```python
app_code ='''


import gdown
import os

if not os.path.exists('resume_matcher_model.h5'):
    print("Downloading model from Google Drive...")
    file_id = "1sOauUHYz8TfbW-aBXY-z0dkdQrAC59eo"
    url = f"https://drive.google.com/uc?id={file_id}"
    gdown.download(url, 'resume_matcher_model.h5', quiet=False)

import gradio as gr
import zipfile
import tempfile
import shutil
import tensorflow as tf
from transformers import BertTokenizer
from parsers import parse_file

model = tf.keras.models.load_model('resume_matcher_model.h5')
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

def process_zip(zip_file):
    """Main function for Gradio"""
    try:
        temp_dir = tempfile.mkdtemp()
        # Save uploaded zip
        zip_path = os.path.join(temp_dir, "upload.zip")
        with open(zip_path, "wb") as f:
            f.write(zip_file.read())




        extract_path = os.path.join(temp_dir, 'extracted')
        os.makedirs(extract_path, exist_ok=True)


        with zipfile.ZipFile(zip_path, 'r') as z:
            z.extractall(extract_path)


        all_files = []
        for root, dirs, files in os.walk(extract_path):
            for f in files:
                if f.lower().endswith(('.odt', '.pdf', '.docx', '.txt')):
                    all_files.append(os.path.join(root, f))

        if len(all_files) < 2:
            shutil.rmtree(temp_dir, ignore_errors=True)
            return "Need at least 2 files (job and resume)", "", ""

        job_file = None
        resume_file = None

        for file_path in all_files:
            file_name = os.path.basename(file_path).lower()
            if any(keyword in file_name for keyword in ['job', 'position', 'description', 'jd']):
                job_file = file_path
            elif any(keyword in file_name for keyword in ['resume', 'cv', 'curriculum', 'bio']):
                resume_file = file_path

        if not job_file:
            job_file = all_files[0]
        if not resume_file:
            resume_file = all_files[1] if len(all_files) > 1 else all_files[0]


        job_text = parse_file(job_file)
        resume_text = parse_file(resume_file)


        test_encoded = tokenizer(
            text=job_text,
            text_pair=resume_text,
            padding='max_length',
            truncation=True,
            max_length=512,
            return_tensors='tf',
            return_token_type_ids=True

        )

        test_input = {
            'input_ids': test_encoded['input_ids'],
            'attention_mask': test_encoded['attention_mask'],
            'token_type_ids': test_encoded['token_type_ids']
        }


        prediction = model.predict(test_input)
        match_score = float(prediction[0][0])
        percentage = match_score * 100


        shutil.rmtree(temp_dir, ignore_errors=True)


        result_text = f"""
        MATCH RESULT:

        Score: {match_score:.3f}/1.0 ({percentage:.1f}%)

        """

        if match_score > 0.8:
            result_text += "EXCELLENT MATCH!"
        elif match_score > 0.6:
            result_text += "GOOD MATCH"
        elif match_score > 0.4:
            result_text += "AVERAGE MATCH"
        else:
            result_text += "POOR MATCH"

        job_preview = job_text[:200] + "..." if len(job_text) > 200 else job_text
        resume_preview = resume_text[:200] + "..." if len(resume_text) > 200 else resume_text

        return result_text, job_preview, resume_preview

    except Exception as e:
        return f"Error: {str(e)}", "", ""

interface = gr.Interface(
    allow_flagging="never",
    fn=process_zip,
    inputs=gr.File(label="pload ZIP File", file_types=[".zip"]),
    outputs=[
        gr.Textbox(label="Match Result", lines=8),
        gr.Textbox(label="Job Preview", lines=4),
        gr.Textbox(label="Resume Preview", lines=4)
    ],
    title="AI Resume Matcher",
    description="Upload a ZIP file containing both job description and resume files (.odt, .pdf, .docx, .txt)",
    theme="soft",
    examples=[["sample.zip"]] if os.path.exists("sample.zip") else None
)


css = """
.gradio-container {
    max-width: 800px !important;
}
.result-box {
    background: linear-gradient(135deg, #667eea 0%, #764ba2 100%);
    color: white;
    padding: 20px;
    border-radius: 10px;
    margin: 10px 0;
}
"""

if __name__ == "__main__":
    interface.launch()
'''

with open('app.py', 'w') as f:
    f.write(app_code)

print("app.py created")
```
